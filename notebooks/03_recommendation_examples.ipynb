Side-by-side top-5 recommendations for three example users, one per method, so the qualitative differences between algorithms are visible. The table is generated by `scripts/recommendation_examples.py` (single source of truth, also used by the report).

```python
%run ../scripts/recommendation_examples.py
```

```python
import pandas as pd
from recsys.config import PROCESSED_DIR
ex = pd.read_csv(PROCESSED_DIR.parents[1] / 'docs' / 'recommendation_examples.csv')
for uid in ex['user_id'].unique():
    sub = ex[ex['user_id'] == uid]
    print(f"\nUser {uid} listens to: {sub['user_taste'].iloc[0]}")
    display(sub[['method','recommendations']].set_index('method'))
```

## What the examples show

- **Popularity is taste-blind.** It serves the same mainstream pop (Britney Spears, Katy Perry, Lady Gaga) to every user - including an ambient / modern-classical listener, where it is wildly off. This is the popularity-bias problem made concrete.
- **The personalised methods align with the user's genre.** For the 80s-synthpop user they converge on Pet Shop Boys, OMD, Eurythmics, Erasure; for the ambient user on Aphex Twin, Biosphere, Alva Noto.
- **Methods differ in character.** Item-item CF stays closest to the user's exact niche (even obscure artists); content-based pulls tag-similar artists; social surfaces what friends play (e.g. Depeche Mode's own members for a Depeche Mode fan); ALS generalises and can drift toward the mainstream for very niche tastes.
- This is the qualitative counterpart to the quantitative comparison in notebook 02.
